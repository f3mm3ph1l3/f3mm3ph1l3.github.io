## Probability of Encounter
To begin to shed some light on this *general trend*, we do need to understand something about why men are attracted to *trans women*. In the [previous article](https://femmephile.com/blog/2024/08/29/heterosex-defined) I touch on this a little, but here we will explore it more *in-depth*. In the previous article we suggested that there must be something inherent about *trans women* that would make them attractive to heterosexual men. While trans women are feminine, and often indistinguishable from *cis women* (or biological females), there must be more to it as we suggested in the article: if it is only *femininity* that attracts *heterosexual males* why not choose a *biological female*?

First let's be clear about the facts: *transgender* people in the *United States of America* represent somewhere between an upper estimate of $1.14\%$ to a more *conservative* estimate of $0.5\%$ (according to [USAFacts](https://usafacts.org/articles/what-percentage-of-the-us-population-is-transgender/) and the [Williams Institutue](https://williamsinstitute.law.ucla.edu/wp-content/uploads/Trans-Pop-Update-Jun-2022.pdf) respectively). Understand that this is **only** the percent of *transgender people* that make up the USA population. The *Williams Institute* determined that of that $0.5\%$ of the US population that are transgender, only $38\%$ are *trans women*. This lowers the percent further to $0.19\%$ of the US population. So in the US you will encounter about $1.9$ trans women per every $1,000$ people:

$$
\frac{\text{1.9 Trans Women}}{\text{1000 People}}
$$

Conversely *cis women* (i.e. biological females) make up about $50\%$ of the US population:

$$
\frac{\text{500 Cis Women}}{\text{1000 People}}
$$

And finally we have the *encounter frequency* of *trans* to *cis* women:

$$
\frac{\text{1.9 Trans Women}}{\text{500 Cis Women}}
$$

So you can, on average, expect that *heterosexual men* who are **randomly** meeting women, to meet almost $2$ trans women for every $500$ cis women. The point that I am making is simply: *heterosexual men* are very **unlikely** to meet *trans women* **unintentionally**. They must **actively** pursue them. Now that doesn't mean that there are not cases when your *average heterosexual male* stumbles across a *trans woman* and they develop a relationship, it's just **more likely** for him to meet a *cis woman* that he will match with: the competition is **steep**.

## Conflicting Interests
Where the conflict seems to arise, is specifically for trans women who are *uncomfortable* with the *"male aspects"* of their person (i.e. *gender dysphoria*). These trans women often decide to undergo *sex-reassignment* surgery, out of necessity to deal with their dysphoria, but in creating an *artificial vagina* they are now becoming more **female** (not necessarily more *feminine*). 

As I have mentioned previously, this may actually *undermine* their appeal to the average *heterosexual male* interested in pursuing them sexually/romantically. Certainly there will be *heterosexual males* who will be interested in **post-op** trans women (i.e. trans women who have had sex-reassignment surgery), I have no doubt about that. But I suspect these post-op trans women will experience a *smaller dating pool* compared to what **pre-op** or **non-op** will experience. And why? Because both pre-op and non-op are *"less female"* (while still being highly feminine) than post-op, making them more *appealing* and *unique* in the dating market.
